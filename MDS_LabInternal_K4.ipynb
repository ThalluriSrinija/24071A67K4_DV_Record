```python
# Pandas DataFrame
data = {
    "Price": [2500000,3000000,3300000,4500000,5000000,5500000],
    "Area": [1000, 1200, 1500, 1800,2000,2500]
}
df = pd.DataFrame(data)
print(df)
```

```python
#dataframes from list
data = [
    [2500000,1000],
    [3000000, 1200],
    [3300000, 1500]
]

df = pd.DataFrame(data, columns=["Price", "Area"])
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/House-Price.csv")
```

```python
#line plot
plt.plot(df["sqft_living"], df["price"])
plt.xlabel("sqft_living")
plt.ylabel("price")
plt.title("eLine Plot")
plt.show()
```

```python
#bar plot
plt.bar(df["sqft_living"], df["price"])
plt.xlabel("sqft_living")
plt.ylabel("price")
plt.show()
```

```python
#histogram
plt.hist(df["price"])
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
#density plot
df["price"].plot(kind="density")
plt.show()
```

```python
#scatter plot
plt.scatter(df["sqft_living"], df["price"])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("House Area vs Price")
plt.show()
```
