```python
# Access an element using one-dimensional indexing
import numpy as np

arr = np.array([114, 20, 30, 40, 50])

print(arr[0])
print(arr[3])
```

```python
# Access an element using two-dimensional indexing
import numpy as np

arr = np.array([[114, 20, 30],
                [40, 50, 60]])

print(arr[0, 0])
print(arr[1, 2])
```

```python
# Access an element using three-dimensional indexing
import numpy as np

arr = np.array([
    [[114, 20], [30, 40]],
    [[50, 60], [70, 80]]
])

print(arr[0, 0, 0])
print(arr[1, 1, 1])
```

```python
# Extract elements using one-dimensional slicing
import numpy as np

arr = np.array([114, 20, 30, 40, 50, 60])

print(arr[1:5])
print(arr[:4])
print(arr[2:])
```

```python
# Extract elements using two-dimensional slicing
import numpy as np

arr = np.array([[114, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print(arr[:, 1:])
```

```python
# Extract elements using slicing with a step value
import numpy as np

arr = np.array([114, 20, 30, 40, 50, 60, 70])

print(arr[::2])
```

```python
# Reshape a one-dimensional array into a two-dimensional array
import numpy as np

arr = np.array([114, 20, 30, 40, 50, 60])

new_arr = arr.reshape(2, 3)

print(new_arr)
```

```python
# Reshape a one-dimensional array into a three-dimensional array
import numpy as np

arr = np.array([114, 20, 30, 40, 50, 60, 70, 80])

new_arr = arr.reshape(2, 2, 2)

print(new_arr)
```

```python
# Convert a multi-dimensional array into a one-dimensional array
import numpy as np

arr = np.array([[114, 20],
                [30, 40]])

print(arr.flatten())
```

```python
# Join arrays using concatenate
import numpy as np

arr1 = np.array([114, 20, 30])
arr2 = np.array([40, 50, 60])

result = np.concatenate((arr1, arr2))

print(result)
```

```python
# Join arrays vertically using stack
import numpy as np

arr1 = np.array([114, 20, 30])
arr2 = np.array([40, 50, 60])

result = np.stack((arr1, arr2))

print(result)
```

```python
# Join arrays using vertical stack
import numpy as np

arr1 = np.array([114, 20, 30])
arr2 = np.array([40, 50, 60])

result = np.vstack((arr1, arr2))

print(result)
```

```python
# Join arrays using horizontal stack
import numpy as np

arr1 = np.array([114, 20, 30])
arr2 = np.array([40, 50, 60])

result = np.hstack((arr1, arr2))

print(result)
```

```python
# Join arrays using depth stack
import numpy as np

arr1 = np.array([[114, 20]])
arr2 = np.array([[30, 40]])

result = np.dstack((arr1, arr2))

print(result)
```

```python
# Split an array into equal parts
import numpy as np

arr = np.array([114, 20, 30, 40, 50, 60])

result = np.split(arr, 3)

print(result)
```

```python
# Split an array into unequal parts
import numpy as np

arr = np.array([114, 20, 30, 40, 50])

result = np.array_split(arr, 3)

print(result)
```

```python
# Split a two-dimensional array horizontally
import numpy as np

arr = np.array([[114, 20, 30, 40],
                [50, 60, 70, 80]])

result = np.hsplit(arr, 2)

print(result)
```

```python
# Split a two-dimensional array vertically
import numpy as np

arr = np.array([[114, 20],
                [30, 40],
                [50, 60],
                [70, 80]])

result = np.vsplit(arr, 2)

print(result)
```
