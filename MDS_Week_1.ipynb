## Performance Comparison: Python Lists vs. NumPy Arrays

This section compares the time efficiency of fundamental operations using standard Python lists and NumPy arrays. We will look at creating large data structures, summing their elements, and performing element-wise multiplication.

```python
import time

# Define a large number of elements for comparison
n_elements = 10**7
```

### Python List Operations

```python
print(f"--- Python List Operations ({n_elements} elements) ---")

# List Creation
start_time = time.time()
py_list = list(range(n_elements))
end_time = time.time()
print(f"List Creation Time: {end_time - start_time:.4f} seconds")

# List Summation
start_time = time.time()
_ = sum(py_list)
end_time = time.time()
print(f"List Summation Time: {end_time - start_time:.4f} seconds")

# List Element-wise Multiplication
start_time = time.time()
# Using a list comprehension for element-wise multiplication
_ = [x * 2 for x in py_list]
end_time = time.time()
print(f"List Element-wise Multiplication Time: {end_time - start_time:.4f} seconds")
```

### NumPy Array Operations

```python
print(f"\n--- NumPy Array Operations ({n_elements} elements) ---")

# NumPy Array Creation
start_time = time.time()
np_array = np.arange(n_elements)
end_time = time.time()
print(f"NumPy Array Creation Time: {end_time - start_time:.4f} seconds")

# NumPy Array Summation
start_time = time.time()
_ = np_array.sum()
end_time = time.time()
print(f"NumPy Array Summation Time: {end_time - start_time:.4f} seconds")

# NumPy Array Element-wise Multiplication
start_time = time.time()
_ = np_array * 2
end_time = time.time()
print(f"NumPy Array Element-wise Multiplication Time: {end_time - start_time:.4f} seconds")
```

### Observations

As you can see from the execution times, NumPy arrays are significantly faster than standard Python lists for numerical operations, especially with a large number of elements. This is due to NumPy's optimized C implementations and efficient memory management.

### Memory Usage Comparison: Python Lists vs. NumPy Arrays

Beyond time efficiency, memory consumption is another crucial aspect where NumPy arrays often outperform Python lists. Let's compare their memory footprints for the same number of elements.

```python
import sys

# Calculate memory usage for Python list
# sys.getsizeof returns the size of the list object itself, plus a pointer to each element.
# We need to account for the size of each element if they are Python integers.
list_memory = sys.getsizeof(py_list)
element_size = sys.getsizeof(py_list[0]) if py_list else 0 # Get size of one element
total_list_elements_memory = sum(sys.getsizeof(item) for item in py_list)
# A more accurate estimation for the list containing Python integers:
# size of list object itself + (number of elements * size of reference) + (number of elements * size of each integer object)
# However, `total_list_elements_memory` below is simpler and often used for comparison.

print(f"\n--- Memory Usage ({n_elements} elements) ---")
print(f"Python List Memory (list object only): {list_memory} bytes")
print(f"Python List Total Memory (list object + elements): {total_list_elements_memory} bytes")

# Calculate memory usage for NumPy array
# NumPy's .nbytes attribute gives the total bytes consumed by the array elements.
np_array_memory = np_array.nbytes
print(f"NumPy Array Memory: {np_array_memory} bytes")
```

### Observations on Memory

NumPy arrays typically use significantly less memory than Python lists for storing numerical data. This is because NumPy stores elements in a contiguous block of memory and uses a fixed-size data type (e.g., `int64`), whereas Python lists store pointers to individual Python objects (each with its own overhead) that can be of varying types.
